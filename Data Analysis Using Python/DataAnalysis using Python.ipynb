```python
#Cleaning missing values
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['age']=df['age'].fillna(df['age'].mean()) # filling missing values in age column with average age
```

```python
df.isnull().sum()
```

```python
df['gender']=df['gender'].fillna(df['gender'].mode()[0])#filling missing values in gender column with mode
df['likes']=df['likes'].fillna(df['likes'].mean())#likes column with average likes
df['comments']=df['comments'].fillna(df['comments'].mean())#comments with avg no.of comments
df['shares']=df['shares'].fillna(df['shares'].median())#shares with median
df = df.dropna(subset=['sentiment'])#removing rows with missing sentiment
```

```python
df.isnull().sum()
```

```python
df
```

```python
#duplicate handling
```

```python
df.duplicated().sum()#to summarise duplicate entries
df = df.drop_duplicates()# removing it
```

```python
#counting
def count_hashtags(x):
    hashtags = str(x).split()
    return len([word for word in hashtags if word.startswith('#')])
df['hashtag_count'] = df['hashtags'].fillna('').apply(count_hashtags)
print(df[['hashtags','hashtag_count']].head(10))
```
